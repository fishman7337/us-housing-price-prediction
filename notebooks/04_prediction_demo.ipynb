<h1>Prediction Demo and Deployment Considerations</h1>
<hr/>
<p>This notebook demonstrates how the trained model can be loaded and used for prediction. It also documents the practical implications and limitations of using a housing price prediction model.</p>


<hr/>
<h1>1.&nbsp;&nbsp;&nbsp;&nbsp;Importing Required Modules</h1>
<hr/>
<p>The saved model is loaded using the project package. If the model file does not exist, the notebook trains a model first so that the prediction demonstration can still run.</p>


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

src_path = PROJECT_ROOT / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
```

```python
import pandas as pd

from us_housing_price_prediction.config import DEFAULT_MODEL_PATH
from us_housing_price_prediction.modeling import load_model, save_model, train_and_evaluate
```

<hr/>
<h1>2.&nbsp;&nbsp;&nbsp;&nbsp;Loading the Saved Model</h1>
<hr/>
<p>The final trained pipeline includes feature engineering, preprocessing, categorical encoding, and regression modeling. Therefore, new prediction data should be provided in the same raw feature format as the training data.</p>


```python
if not DEFAULT_MODEL_PATH.exists():
    model, _ = train_and_evaluate()
    save_model(model, DEFAULT_MODEL_PATH)

model = load_model(DEFAULT_MODEL_PATH)
print(f"Loaded model from: {DEFAULT_MODEL_PATH}")
```

<p>With reference to the output above, the saved model pipeline has been loaded successfully and is ready for prediction.</p>


<hr/>
<h1>3.&nbsp;&nbsp;&nbsp;&nbsp;Creating a Sample House Record</h1>
<hr/>
<p>To generate a prediction, a new house record must contain the same input fields used by the model. The target price is not included because that is what the model is trying to predict.</p>


```python
sample_house = pd.DataFrame(
    [
        {
            "city": "Chicago",
            "house_area_sqm": 742.0,
            "bedrooms": 4,
            "toilets": 2,
            "stories": 3,
            "renovation_status": "furnished",
        }
    ]
)
sample_house
```

<p>The sample record contains the required fields. Since the pipeline performs preprocessing internally, the categorical values can remain as normal text values.</p>


<hr/>
<h1>4.&nbsp;&nbsp;&nbsp;&nbsp;Predicting House Price</h1>
<hr/>
<p>The model now predicts the estimated housing price for the sample record. The result should be interpreted as an estimate rather than a guaranteed market value.</p>


```python
prediction = float(model.predict(sample_house)[0])
print(f"Predicted price: ${prediction:,.2f}")
```

<hr/>
<h1>5.&nbsp;&nbsp;&nbsp;&nbsp;Deployment and Practical Implications</h1>
<hr/>
<p>The model can support prototype real estate analytics by providing quick estimated prices from structured housing attributes. However, real deployment would require stronger infrastructure and governance.</p>
<br/>
<ul>
    <li><strong>Data quality:</strong> Predictions are only as reliable as the training data.</li>
    <li><strong>Market changes:</strong> Housing prices can change due to economic and location-specific factors not present in this dataset.</li>
    <li><strong>User impact:</strong> The prediction should support human judgment rather than replace it.</li>
    <li><strong>Ethics:</strong> The model should not be used as the sole basis for lending, taxation, insurance, or legal valuation decisions.</li>
</ul>
