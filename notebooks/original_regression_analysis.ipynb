<h3>4.3.1&nbsp;&nbsp;&nbsp;&nbsp;Identifying Outliers Using Boxplot</h3>

<p>The easiest and most used method to visualise outlier is using a Boxplot. A Boxplot will contain a few important parts such as the minimum, maximum, lower quartile, second quartile (median), and upper quartile. With all these parts present, I am able to perform and visualise important calculations such as the Inter-Quartile Range which will assist us in establishing the lower and upper fence should there be any Outliers. An Outlier will be flagged if it is below the Minimum or above the Maximum. Through the Boxplot, I am also able to visualise how many Outliers is present within a Numerical Category. As 'City' and 'Renovation Status' are represented by Dummy Variable, they will be excluded from this operation.</p>

<h2>4.4&nbsp;&nbsp;&nbsp;&nbsp;Visualising Outliers Ratio</h2>

<p>To see the amount of Outliers and the scale of it compared to my overall Dataset. This will allow me to decide whether I should remove or how should I addressed the Outliers within my dataset without affecting the scope of the dataset.</p>

```python
# Create a copy of the DataFrame to label outliers
df_labeled = df.copy()

# List of numeric columns to analyze
numeric_cols = ['House Area (sqm)', 'No. of Bedrooms', 'No. of Toilets', 'Stories', 'Price ($)']

# Add a new column to indicate whether each row is an outlier or not
df_labeled['Outlier'] = 'Non-Outlier'

# Identify outliers using the IQR method
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Label rows as 'Outlier' if they fall outside the IQR bounds
    df_labeled.loc[(df[col] < lower_bound) | (df[col] > upper_bound), 'Outlier'] = 'Outlier'

# Count the number of outliers and non-outliers
outlier_counts = df_labeled['Outlier'].value_counts()

# Plotting the pie chart for the ratio of outliers to non-outliers
plt.figure(figsize=(8, 6))
plt.pie(outlier_counts, labels=outlier_counts.index, autopct='%1.1f%%', startangle=90, colors=['yellowgreen', 'salmon'])
plt.title('Ratio of Outliers to Non-Outliers')
plt.show()
```

<h2>4.6&nbsp;&nbsp;&nbsp;&nbsp;Creation of New Features</h2>

In this sub-section, I will be creating 2 new features which I believe will make the model more accurate and be beneficial. During my research, I stumbled upon 4 formulas which can be used to create 4 new features. I will be adding House Area per Bedroom, Toilet to Bedroom Ratio, Total Rooms and Stories House Area Intersection. I will be utilising the following formulas to calculate the New Features.
<br/>
$$
\text{House Area per Bedroom} = \frac{\text{House Area (sqm)}}{\text{No. of Bedrooms}}
$$
<br/>
$$
\text{Toilet to Bedroom Ratio} = \frac{\text{No. of Toilets}}{\text{No. of Bedrooms}}
$$
<br/>
$$
\text{Total Rooms} = \text{No. of Toilets} + \text{No. of Bedrooms}
$$
<br/>
$$
\text{Stories House Area Intersection} = \text{Stories} \times \text{House Area (sqm)}
$$
<br/>
The formulas are also capable of describing the relationship between the features that are used to calculate the new features:

- **House Area per Bedroom**:
    - If House Area **increases** while No. of Bedrooms **remains the same**, House Area per Bedroom will **increase**.
    - If House Area **remains the same** while No. of Bedrooms **increases**, House Area per Bedroomt will **decrease**.

- **Toilet to Bedroom Ratio**:
    - If No. of Toilets **increases** while No. of Bedrooms **remains the same**, Toilet to Bedroom Ratio will **increase**.
    - If No. of Toilets **remains the same** while No. of Bedrooms **increases**, Toilet to Bedroom Ratio will **decrease**.

- **Total Rooms**:
    - If No. of Toilets **increases** while No. of Bedrooms **remains the same**, Total Rooms will **increase**.
    - If No. of Toilets **remains the same** while No. of Bedrooms **increases**, Total Rooms will **increase**.

- **Stories House Area Intersection**:
    - If Stories **increases** while House Area (sqm) **remains the same**, Stories House Area Intersection will **increase**.
    - If Stories **remains the same** while House Area (sqm) **increases**, Stories House Area Intersection will **increase**.

Upon stating the formula and the relationship, I will proceed with creating the new features in the cell below.

```python
# Creating House Area per Bedroom Feature
df['House_Area_per_Bedroom'] = df['House Area (sqm)'] / df['No. of Bedrooms']

# Creating Toilet to Bedroom Ratio Feature
df['Toilet_to_Bedroom_Ratio'] = df['No. of Toilets'] / df['No. of Bedrooms']

# Creating Total Rooms Feature
df['Total_Rooms'] = df['No. of Bedrooms'] + df['No. of Toilets']

# Creating Stories House Area Interaction Feature
df['Stories_House_Area_Interaction'] = df['Stories'] * df['House Area (sqm)']

# Viewing of Updated Data Table
df.head(10).style.background_gradient(cmap="Blues")
```

<h2>8.1&nbsp;&nbsp;&nbsp;&nbsp;Learning Curve Plotting Function</h2>

<p>I will be plotting the learning curve multiple times to detect for overfitting and underfitting or ideal fitting. Therefore, I will be utilising a function such that I do not have to retype the code everytime when plottng the learning curve. A learning curve is capable of showing the following results and the action plan that needs to be done to rectify the issue (if any).</p>
<br/>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>Cross-Validation Score matches the Training Score at the final point.</li>
    <li>Hyperparameter have been tuned successfully.</li>
</ul>
<p><strong><u>Overfitting</u></strong></p>
<ul>
    <li>Cross-Validation Score fall short of the Training Score.</li>
    <li>Hyperparameter needs to be adjusted accordingly by reducing the model's complexity.</li>
</ul>
<p><strong><u>Underfitting</u></strong></p>
<ul>
    <li>Training Score fall short of the Cross-Validation Score.</li>
    <li>Hyperparameter needs to be adjusted accordingly by increase the model's complexity.</li>
</ul>
<br/>
<p>With the potential results and action plan recorded, I will proceed to implement the Learning Curve Plotting function in the cell below.</p>

```python
# Learning Curve Function for Regression Subplot
def plot_learning_curve_subplot(estimator, title, X, y, ax, ylim=None, cv=None, n_jobs=1, train_sizes=np.linspace(0.1, 1.0, 5)):
    sns.set(style="whitegrid")
    ax.set_title(title, fontsize=12)
    ax.set_xlabel("Training examples", fontsize=10)
    ax.set_ylabel("R^2 Score", fontsize=10)
    if ylim is not None:
        ax.set_ylim(*ylim)

    # Compute learning curve data
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, y, cv=cv, n_jobs=n_jobs, train_sizes=train_sizes, scoring='r2'
    )
    
    # Calculate mean and standard deviation
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)

    # Fill between for train and test standard deviations
    ax.fill_between(train_sizes, train_scores_mean - train_scores_std,
                    train_scores_mean + train_scores_std, alpha=0.1, color="blue")
    ax.fill_between(train_sizes, test_scores_mean - test_scores_std,
                    test_scores_mean + test_scores_std, alpha=0.1, color="orange")

    # Plot the mean scores for training and testing sets
    sns.lineplot(x=train_sizes, y=train_scores_mean, color="blue", marker='o', label="Training Score", ax=ax)
    sns.lineplot(x=train_sizes, y=test_scores_mean, color="orange", marker='o', label="Cross-Validation Score", ax=ax)

    # Highlight the final points for emphasis
    ax.scatter(train_sizes[-1], train_scores_mean[-1], color='blue', s=50, edgecolor='k', linewidth=1.5, label='Final Training Point')
    ax.scatter(train_sizes[-1], test_scores_mean[-1], color='orange', s=50, edgecolor='k', linewidth=1.5, label='Final Cross-Validation Point')
    ax.legend(loc="best", fontsize=8)
    ax.grid(True)

# Code Execution Confirmation
print("Function Successfully Defined")
```

<h2>8.4&nbsp;&nbsp;&nbsp;&nbsp;Scoring Evaluation for Models</h2>

<p>The last part of the Preliminary Training of Models is to evaluate the respective scoring of the model. These scoring includes Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE) and R2 Score. I will also be colour coding the data which are not ideal and data that are ideal in red and green respectively. After the scoring evaluation, I may need to compare certain models with each other to evaluate their individual performance.</p>

```python
# Define scoring metrics for cross-validation
scoring_metrics = {
    'MAE': make_scorer(mean_absolute_error),
    'MSE': make_scorer(mean_squared_error),
    'RMSE': make_scorer(lambda y_true, y_pred: mean_squared_error(y_true, y_pred, squared=False)),
    'R2': make_scorer(r2_score),
    'MAPE': make_scorer(mean_absolute_percentage_error),
    'EVS': make_scorer(explained_variance_score)
}

# List to hold cross-validation results for all models
results = []

# Perform cross-validation on each model and store the results
for model_name, model in regression_models:
    cv = ShuffleSplit(n_splits=10, test_size=0.2, random_state=28)
    scores = cross_validate(
        model,
        X_train,
        Y_train,
        scoring=scoring_metrics,
        cv=cv,
        return_train_score=True,
        n_jobs=-1
    )
    
    # Append the average score for each model to the results
    avg_scores = {metric: scores[metric].mean() for metric in scores}
    avg_scores['model'] = model_name
    results.append(avg_scores)

# Convert the list of results to a DataFrame, setting the model names as index
outputs_df = pd.DataFrame(results)
outputs_df.set_index('model', inplace=True)

# Function to highlight thresholds for every single data box
def highlight_thresholds(df):
    # Function to apply color based on value range
    def apply_highlight(val, good_thresh, acceptable_thresh):
        if val < acceptable_thresh:
            return 'background-color: red; color: white'
        elif val < good_thresh:
            return 'background-color: yellow; color: black'
        else:
            return 'background-color: green; color: white'

    # Apply styling for various metrics using thresholds
    styled_df = df.style.applymap(
        lambda x: apply_highlight(x, 50000, 100000), subset=['test_MAE', 'train_MAE']
    ).applymap(
        lambda x: apply_highlight(x, 1e10, 2e10), subset=['test_MSE', 'train_MSE', 'test_RMSE', 'train_RMSE']
    ).applymap(
        lambda x: apply_highlight(x, 0.6, 0.4), subset=['test_R2', 'train_R2', 'test_EVS', 'train_EVS']
    ).applymap(
        lambda x: apply_highlight(x, 0.1, 0.2), subset=['test_MAPE', 'train_MAPE']
    )

    return styled_df

# Apply the new highlighting to your DataFrame
highlighted_df = highlight_thresholds(outputs_df)

# Display the highlighted DataFrame
highlighted_df
```

<h3>9.1.3&nbsp;&nbsp;&nbsp;&nbsp;Random Forest Regressor's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the Random Forest Regressor, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Random Forest Regressor. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_rfr = {
    'randForestRegressor__n_estimators': [200, 250],
    'randForestRegressor__max_depth': [10, 12],
    'randForestRegressor__min_samples_split': [10, 15],
    'randForestRegressor__min_samples_leaf': [5, 10],
    'randForestRegressor__max_features': ['sqrt'],
    'randForestRegressor__bootstrap': [True]
}

# Create a GridSearchCV instance using KFold
grid_search_rfr = GridSearchCV(
    estimator=pipeline_rfr,
    param_grid=grid_param_rfr,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_rfr.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_rfr.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1

# Highlight the top 3 results in green
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(
    lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1
).set_properties(**{'border': '1px solid black'})

# Display the highlighted DataFrame
display(highlighted_grid_sorted_results_df)
```

<h3>9.2.1&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree's Initiation and Pipeline</h3>

<p>In order to start training the data for Decision Tree, I will firstly initiate it and also set up the pipeline which will scale the training data. Scaling of data will also reduce biasness, ensure the feature is dimensionless and improve the model more accurate and effective as extreme values are removed.</p>

```python
# Initiating Decision Tree
dt = DecisionTreeRegressor(random_state=28)

# Define the pipeline with scaling
pipeline_dt = Pipeline([
    ('scaler', StandardScaler()),
    ('decisionTreeRegressor', dt)
])

# Code Execution Confirmation
print("Decision Tree Successfully Initated and Pipeline Established")
```

<p>I will be using the KFold to conduct the cross-validation strategy within the GridSearchCV and RandomizedSearchCV to find the best Hyperparameter for the Decision Tree. KFold is also capable of reducing variance and bias which will directly improve the model's performance in terms of its fitting.</p>

```python
# Defining KFold Cross-Validator
kf = KFold(n_splits = 10, shuffle = True, random_state = 28)

# Code Execution Confirmation
print("KFold have been defined successfully")
```

<h3>9.2.3&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best Hyperparameter for the Decision Tree, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Decision Tree. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my Decision Tree, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_dt = {
    'decisionTreeRegressor__max_depth': [4, 5],
    'decisionTreeRegressor__min_samples_leaf': [70, 75],
    'decisionTreeRegressor__ccp_alpha': [0.000005, 0.00001],
    'decisionTreeRegressor__min_samples_split': [150],
    'decisionTreeRegressor__max_leaf_nodes': [50, 55]
}

# Create a GridSearchCV instance using KFold
grid_search_dt = GridSearchCV(
    estimator = pipeline_dt,
    param_grid = grid_param_dt,
    cv = kf,
    verbose = 1,
    n_jobs = -1
)

# Fit GridSearchCV to the training data
grid_search_dt.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_dt.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

<h3>10.3.3&nbsp;&nbsp;&nbsp;&nbsp;Linear Support Vector Regression Regression's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best Hyperparameter for the Linear Support Vector Regression Regression, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Linear Regression. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid to search for GridSearchCV
grid_param_svr = {
    'svr__C': np.logspace(-3, 3, 15),
    'svr__epsilon': np.linspace(0.1, 1.0, 10),
}

# Create a GridSearchCV instance
grid_search_svr = GridSearchCV(
    estimator=pipeline_svr,
    param_grid=grid_param_svr,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_svr.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_svr.cv_results_)

grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter distribution to search
param_dist_svr = {
    'svr__C': np.linspace(0.001, 20, 100),
    'svr__epsilon': np.linspace(0.1, 1.0, 10),
}

# Create a RandomizedSearchCV instance
random_search_svr = RandomizedSearchCV(
    estimator=pipeline_svr,
    param_distributions=param_dist_svr,
    n_iter=10,
    scoring='r2',
    cv=kf,
    random_state=28,
    n_jobs=-1,
    verbose=2
)

# Fit RandomizedSearchCV to the training data
random_search_svr.fit(X_train, Y_train)

# Create a DataFrame to display the results
random_results_df = pd.DataFrame(random_search_svr.cv_results_)
random_sorted_results_df = random_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
random_sorted_results_df.index += 1
highlighted_random_sorted_results_df = random_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_random_sorted_results_df)
```

<h3>10.4.3&nbsp;&nbsp;&nbsp;&nbsp;Gradient Boosting Regressor's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best Hyperparameter for the Gradient Boosting Regressor, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Linear Regression. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid to search for GridSearchCV
grid_param_gbr = {
    'gbr__n_estimators': [500, 800, 1000],
    'gbr__learning_rate': [0.001, 0.005, 0.01],
    'gbr__max_depth': [3, 4],
    'gbr__min_samples_split': [30, 50, 70],
    'gbr__min_samples_leaf': [15, 20, 30],
    'gbr__subsample': [0.5, 0.6, 0.7],
    'gbr__max_features': ['sqrt', 'log2', 0.5],
    'gbr__loss': ['ls', 'huber']
}

# Create a GridSearchCV instance
grid_search_gbr = GridSearchCV(
    estimator=pipeline_gbr,
    param_grid=grid_param_gbr,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_gbr.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_gbr.cv_results_)
grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

<h3>10.5.3&nbsp;&nbsp;&nbsp;&nbsp;MLP Regressor's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the MLP Regressor, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the MLP Regressor. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid to search for GridSearchCV
grid_param_mlp = {
    'mlp__hidden_layer_sizes': [(20,), (50,), (20, 20)],
    'mlp__activation': ['relu', 'tanh'],
    'mlp__alpha': [1e-5, 1e-4, 1e-3],
    'mlp__learning_rate_init': [0.001, 0.01],
    'mlp__solver': ['adam', 'sgd'],
    'mlp__learning_rate': ['constant', 'adaptive'],
    'mlp__max_iter': [500, 1000],
}

# Create a GridSearchCV instance
grid_search_mlp = GridSearchCV(
    estimator=pipeline_mlp,
    param_grid=grid_param_mlp,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_mlp.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_mlp.cv_results_)

grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

<p>In order to find the best hyperparameter for the MLP Regressor, I will also be running RandomizedSearchCV which will search through the parameters randomly, depending on the amount of folds. RandomizedSearchCV is a lot more efficient in finding the Best Hyperparameter for the MLP Regressor as it has lesser amount of total fits as compared to GridSearchCV. <p><strong><u>Accuracy</u></strong></p>

Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter distribution to search
param_dist_mlp = {
    'mlp__hidden_layer_sizes': [(10,), (20,), (10, 10), (20, 20)],
    'mlp__activation': ['relu', 'tanh'],
    'mlp__alpha': np.logspace(-5, -2, 10),
    'mlp__learning_rate': ['constant', 'adaptive'],
    'mlp__learning_rate_init': [0.001, 0.01],
    'mlp__solver': ['adam', 'sgd'],
    'mlp__max_iter': [500, 1000],
}

# Create a RandomizedSearchCV instance
random_search_mlp = RandomizedSearchCV(
    estimator=pipeline_mlp,
    param_distributions=param_dist_mlp,
    n_iter=10,  # Number of parameter settings sampled
    scoring='r2',
    cv=kf,       # 5-fold cross-validation
    random_state=28,
    n_jobs=-1,  # Use all available cores
    verbose=2
)

# Fit RandomizedSearchCV to the training data
random_search_mlp.fit(X_train, Y_train)

# Create a DataFrame to display the results
random_results_df = pd.DataFrame(random_search_mlp.cv_results_)
random_sorted_results_df = random_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
random_sorted_results_df.index += 1
highlighted_random_sorted_results_df = random_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_random_sorted_results_df)
```

GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good R2 score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid to search for GridSearchCV
grid_param_knn = {
    'knn__n_neighbors': [3, 5, 7, 9, 15, 21],    # Larger range to test smoothing
    'knn__weights': ['uniform', 'distance'],     # Test both weighting options
    'knn__algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute'],  # Keep all algorithms
    'knn__leaf_size': [5, 10, 20, 30, 40],       # Test smaller and larger leaf sizes
    'knn__p': [1, 2]
}

# Create a GridSearchCV instance
grid_search_knn = GridSearchCV(
    estimator=pipeline_knn,
    param_grid=grid_param_knn,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_knn.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_knn.cv_results_)

grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

<h1>11.&nbsp;&nbsp;&nbsp;&nbsp;Voting Regressor</h1>

<p>Voting Regressor is an ensemble method which combines multiple Regression models to improve the overall performance and accuracy of the Regression operation. Voting Regressor use the strength of each model in conjunction with each other to come up with the most accurate prediction possible. However, there will definitely still be incorrect predictions despite the improved performance. Voting Regressor will take in the predicted value from each Regressor and calculate the total sum. With respect to each of the weight of the models, the Voting Regressor will then average the total predicted value out and returning the avearaged value as the prediction.</p>

<h2>11.11&nbsp;&nbsp;&nbsp;&nbsp;Cook's Distance Plot</h2>

<p>With the help of Cook's Distance Plot, I am able to see the influential points (Potential Outliers) and subsequently remove them. I can then evaluate the Voting Regressor's performance with and without the Outliers.</p>

```python
# Fit the Voting Regressor to the training data
best_voting_regressor.fit(X_train, Y_train)

# Predict on the training set to get residuals
y_train_pred = best_voting_regressor.predict(X_train)
residuals = Y_train - y_train_pred

# Add a constant to the features
X_train_const = sm.add_constant(X_train)

# Fit an OLS model using statsmodels to get leverage values
ols_model = sm.OLS(Y_train, X_train_const).fit()
influence = ols_model.get_influence()

# Get leverage and Cook's distance
leverage = influence.hat_matrix_diag
cooks_d = influence.cooks_distance[0]

# Plot Cook's Distance
plt.figure(figsize=(10, 6))
plt.stem(np.arange(len(cooks_d)), cooks_d, markerfmt=",")  # Removed use_line_collection
plt.axhline(4 / len(cooks_d), color='red', linestyle='--', lw=1)  # Common threshold
plt.title("Cook's Distance for Voting Regressor")
plt.xlabel('Observation Index')
plt.ylabel("Cook's Distance")
plt.show()

# Identifying potentially influential points
influential_points = np.where(cooks_d > 4 / len(cooks_d))[0]
print(f"Influential points (Cook's Distance > 4/n): {influential_points}")
```

<p>From the Cook's Distance Plot above, I am able to see that there are a few influential points (Outliers) that are above the Threshold of 4. These Outliers will then subsequently be removed and the Voting Regressor will be evaluated on whether it is able to perform better with or without the Outliers.</p>
